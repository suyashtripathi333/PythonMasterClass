```python
# defining a class with instance methods
class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    
    def bark(self):# to access instance var in constructor
        print(f"{self.name} says woof ")
        
dog2 = Dog("Drago",5)
dog2.bark()
        
```

```python
#Modelling a bank acount _>this done by my self 

class Account:
    
    def __init__(self,Name,Age,Nationality,Balance,sign):
        self.Name = Name
        self.Age = Age
        self.Nationality= Nationality
        self.Balance = Balance
        self.sign = sign
    
    def is_eligible(self):
        if self.Age>=18:
            if self.Nationality == "Indian":
                if self.Balance >=2000:
                    print(f"Mr {self.Name} you can have your account with us ")
                else:
                    print(f"Mr {self.Name} your balance is less ")
            else:
                print(f"Mr {self.Name} you are not Indian so sorry")
        else:
            print(f"Mr {self.Name} your age is lees than 18.So not allowed ")
    
    
    def bank_operation(self):
        if self.sign == "-":
            if self.Balance >=5000:
                print(f"Mr {self.Name}  you can withraw the money {self.Balance} ")
                withdraw_amt=int(input("Enter the withdraw amount"))
                self.Balance = self.Balance -withdraw_amt
                print(f"Mr {self.Name} you are left with {self.Balance}")
            else:
                print(f"Mr {self.Name} your balance is low  ")
        
        if self.sign=="+":
            depo_amt = int(input("Enter the amount to deposit"))
            self.Balance = self.Balance + depo_amt
            print(f"Mr {self.Name} your curr balance {self.Balance} ")
        
        if self.sign == "--":
            pass
```

```python
#User must enter sign to deposite money "+" else "-"
```

```python
user1 = Account("Suyash",23,"Indian",20000,"-")
```

```python
user1.is_eligible()
user1.bank_operation()
```
