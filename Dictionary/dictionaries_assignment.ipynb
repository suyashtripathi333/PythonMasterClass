```python
#Create a dictionary with the first 10 positive integers as keys and their squares as values.
pos_dict = {x:x**2 for x in range(1,11)}
# Print the dictionary.
print(pos_dict)
```

```python
#Print the value of the key 5 
print(pos_dict[5])
# and the keys of the dictionary created in Assignment 1.
for key in pos_dict.keys():
    print(key, end=" ")
```

```python
#Add a new key-value pair (11, 121) to the dictionary created in Assignment 1
pos_dict[11]=121
# and then remove the key-value pair with key 1. 
del pos_dict[1]
# Print the modified dictionary.
print(pos_dict)
```

```python
#Iaterate over the dictionary created in Assignment 1 and print ech key-value pair.

for key,value in pos_dict.items():
    print(f"{key} : {value}")
```

```python
#Create a new dictionary containing the cubes of the first 10 positive integers using a dictionary comprehension.
cube_dict = {x:x**3 for x in range(1,11)}
# Print the new dictionary.
print(cube_dict)
```

```python
#Create a dictionary with the first 5 positive integers as keys and their squares as values. 
dict3 = {x:x**2 for x in range(1,5)}
print(dict3)
# Convert the dictionary to a list of tuples and print it.
lst=[]
for key, value in dict3.items():
    lst.append((key,value))
print(lst)
```

```python
#Create a dictionary with the first 10 positive integers as keys and their squares as values. 
dict4 = {x:x**2 for x in range(1,11)}
print(dict4)
# Create a new dictionary containing only the key-value pairs where the key is even. 
even_dict4 = {key:value  for key,value in dict4.items() if key%2==0}
# Print the new dictionary.
print(even_dict4)
```

```python
#Create a dictionary with the first 5 positive integers as keys and their squares as values.
dict5 = {x:x**2 for x in range(1,6)}
print(dict5)
# Create a new dictionary with keys and values swapped.
swap_dict = {value : key  for key,value in dict5.items()}
 
# Print the new dictionary.
print(swap_dict)
```

```python
#Create a default dictionary where each key has a default value of an empty list.
dict6 = {x:[] for x in range(1,11)}
print(dict6)
# Add some elements to the lists and print the dictionary.
for value in dict6.values():
    value.append([x for x in range(1,5)])
print(dict6)
```
