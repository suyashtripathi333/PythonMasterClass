```python
#creating a dictionary 

student = {"name":"Suyash", "age":23 , "grade":24}
print(student)
```

```python
#we cannot use same key again and again -> it takes only unique key 
#ex

student = {"name":"Suyash", "age":23 , "name":"Suyash"}
print(student)
```

```python
#Accessing  Dictionary Elements 
student = {"name":"Suyash", "age":23 , "grade":"A"}

print(student["grade"])

#2. way default get method 

print(student.get('grade'))
print(student.get('last_name'))
print(student.get("last_name","Not available"))
```

```python
#Modifying dictionary Elements
#Dictionary are mutable , so  you can add , update , delete elements 
print(student)
```

```python
#updating the value of the key age 
student["age"] = 33
print(student)

#Adding new key to the dictionary 
student["address"] = "India"
print(student)
```

```python
#delete the key and value pair 
del student["address"]
print(student)
```

```python
#dictionary methods

#to get the key of the dictionary
key = student.keys()
print(key)

#to get the value of of the dictionary

value = student.values()
print(value)

#to get all the key value pairs 
items=student.items()
print(items)
```

```python
#the importance of making shallow copy

#lets make the copy like normally we do 
student_copy = student

print(student)
print(student_copy)
```

```python
#now lets update any key's value in  this dictionary

student["name"]="Suyash2"
print(student)
print(student_copy)
```

##So here we can see the change we applied has changed in copy also that why we need shallow copy 

```python
#how to make shallow copy 

student_cp = student.copy()
print(student)
print(student_cp)
```

```python
student["name"]= "Suyash"
print(student)
print(student_copy)
print(student_cp)#it is shallow copy 
```

here we can see that at third  dictionary print it is a shallow copy here so on updating name key it did not change in shallow copy

```python
#Nested dictionaries 

students={
    "S1":{"name":"Suyash", "age":23},
    "S2":{"name":"Maya","age":32},
    "S3":{"name":"Dev", "age":42}
}
print(students)
```

```python
#Accessing nested dictionaries element 
print(students["S1"]["name"])
print(students["S2"]["age"])
```

```python
#iterating over nested dictionaries 

for stud_id , student_info in students.items():
    print(f"{stud_id} : {student_info}")
    for key,value in student_info.items():
        print(f"{key} : {value}")
    
    
```

```python
print(student_info)
print(stud_id)
```

```python
#dictionary comprehension
squares = {x:x**2 for x in range(1,11)}
print(squares)
```

```python
#iterated squares
for key, value in squares.items():
    print(f"{key}: {value}")
```

```python
#Condition dictionary comprehension
even_squares = {x:x**2 for x in range(1,11) if x%2==0}
print(even_squares)
```

```python
#Use a dictionary to count element in the list 
lst =[1,2,3,4,3,2,4,6,7,4,3,5,6,4,6,7,8,6]

freq_lst={i:lst.count(i)  for i in lst}
print(freq_lst)
```

```python
#Use for loop to the above question
frequency={}

for num in lst:
    if num in frequency:
        frequency[num]+=1
    else:
        frequency[num]=1
print(frequency)
```

```python
#Merge 2 dictionaries into one

dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}

merge_dict = {**dict1,**dict2}
print(merge_dict)
```
