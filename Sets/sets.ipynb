Set are for the collection of unique items and they do not follow any specific order 

```python
#Create  a Set 

#empty set 
empty_set = set()
print(type(empty_set))
```

```python
#putting value in set 
_set = set([1,2,3,4,46,7,5,6])
print(_set)
```

```python
#basic set operations

#1.Adding and removing elements

_set.add(100)
print(_set)


#2.Remove the elements 
_set.remove(3)
print(_set)



```

```python
#here we have pop method 
print(_set)#here in the set at index 0 there is 1 present 
_set.pop()
print(_set)#after using pop it pop out the index 0 so it is first in first out 
```

```python
#clear all elements 
_set.clear()
print(_set)
```

```python
#Set Membership test 
print(3 in _set)#we are using in to check whether 
print(20 in _set)
```

```python
#Mathematical operations
s1={1,2,3,3,4,5,6,7}
s2={8,9,10,11,12,13,14}
```

```python
s3={1,2,3,3,4,5,6,7,10,8}
s4={8,9,10,5,12,7,14,4,2,}
```

```python
#Union -> both s1 and s2 is a union ,no duplicates allowed 

u_set = s1.union(s2)
print(u_set)
```

```python
#Intersection -> elemnets which are in both the set 

inter_set= s1.intersection(s2)
print(inter_set)# because no element in common
```

```python
#Now using s3 and s4 wwe can see the intersection function
inter_set= s3.intersection(s4)
print(inter_set)

```

```python
# intersection-update-> it take te intersection and put into the first set 
s3.intersection_update(s4)
print(s3)#all the elements of intersection got into s3
```

```python
#difference s1.differnce(s2 ) print the element whi are not in s2 but in s1 only

print(s3.difference(s4))
```

```python
#Symmetric differnce unlike differnece which finds uncommon in s1 related to s2
#symmetric differnce finds  it same for s2 also 
s3.symmetric_difference(s4)
```

```python
#Set Methods
set1={1,2,3}
set2={3,4,5}

# is subset of 
print(set2.issubset(set1))
print(set1.issubset(s2))
```

```python
#counting unique words in set 

txt = "Hello here we are learning about set and this lecture is really helpful regarding set "
words = txt.split()
unique_words=set(words)
print(unique_words)
count = len(unique_words)
print(f"There are total {count} words in the sentence  ")
```
