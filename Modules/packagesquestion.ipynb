```python
#Use the `sys` module to print the Python version currently in use and the command-line arguments passed to the script.
import sys

print(sys.version)

version = sys.version_info
print(version)

print(f"Python {version.major}.{version.minor}.{version.micro}")
```

```python
#Use the `math` module to calculate the greatest common divisor (GCD) of two numbers and the factorial of a number.
import math 

print(math.gcd(5,3))
print(math.factorial(5))
```

```python
#Use the `datetime` module to print the current date, 
# calculate the date 100 days from today, and determine the day of the week for a given date.

from datetime import datetime,timedelta

now = datetime.now()
print(now)
day_100 = now + timedelta(days=100)
day_of_week = day_100.strftime("%A")
print(day_of_week)
print(day_100)
```

```python
#Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.
import random

lst = random.choices(range(1,51),k=10)
print(lst)
```
