Import Modules in python : Modules and Package

In Python modules and packages help organize and reuse code.

```python
#we are importing math to use in-built function in that class 

import math
math.sqrt(16)
```

```python
from math import pi,sqrt
print(sqrt(16))
print(pi)
```

```python
import numpy as np
```

```python
np.array([1,2,3,4,5,6,7])
```

```python
#if want to import every module than 
from math import *
```

```python
#now importing the module that we made 

from package.maths import add

add(5,3)
```

```python
from package.subpackage.mult import multiply
multiply(5,6)
```
