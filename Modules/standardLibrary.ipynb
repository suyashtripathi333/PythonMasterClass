```python
#data Serialization
import json

data={
    'name':'suyash',
    'age':23
    
}



json_str = json.dumps(data)#create this file to json file
print(json_str)
print(type(json_str))

#now converting json to dictionary basck
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
#csv
import csv

with open("sample.csv",mode="w",newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name',"age"])
    writer.writerow(['Krish',32])
    
    
with open("sample.csv",mode="r") as file :
    reader = csv.reader(file)
    for row in reader:
        print(row)
    
```
