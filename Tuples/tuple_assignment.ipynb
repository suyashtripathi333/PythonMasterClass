```python
#Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.

tpl = (1,2,3,4,5,6,7)
```

```python

#function for sum of tpl
def sum_tpl(tpl):
    sum =0
    for i in tpl:
        sum=sum+i
    return sum    
```

```python
#fuction for minimum in tpl
def minimum(tpl):
    m=tpl[0]
    for i in tpl:
        if i<m:
            m=i
    return m
```

```python
#function for maximum in tpl
def maximum(tpl):
    m = tpl[0]
    
    for i in tpl:
        if i>m:
            m=i
            
    return m
```

```python
r_sum= sum_tpl(tpl)
print("The sum of the tpl element is : ",r_sum)

r_min = minimum(tpl)
print("The minimum or smallest element in tpl : ",r_min)

r_max = maximum(tpl)
print("The maximumor largest element in tpl is : ",r_max)
```
