**Question 5:** Write a Python program to calculate the factorial of a number.

```python
# Your code here

fact_num = int(input("Enter the number to find it's factorial"))

prod =1 

while fact_num>0:
    prod = prod * fact_num
    fact_num = fact_num-1

print("The factorial of number is: ",prod)
```

```python
# Your code here

year = int(input("enter the year you want to know it's leap or not"))

if year % 4 == 0:
    if year % 100 != 0 or year % 400 ==0:
        print("this is a leap year")
       
    else:
        print("this is   not a leap year")
else:
    print("this is not  leap year ")       
```
