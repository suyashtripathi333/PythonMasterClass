```python
#Creating a function for temperatur conversion 

#celcius to fehrenheit
def temp(a):
    return (a*(9/5)+32)

def temp_cel(a):
    return (a - 32)*(5/9)
    

```

```python
#taking user input here 
a = int(input("enter the temperature in celcius"))

#calling the function and saving its value in result
result = temp(a)
#printing the result to only two decimal value 
print(f"The temp in fehrenheit is {result:.2f} F ")
```

Password Strength Checker


```python
#to check the password is strong 

def check_points(p):
    point=0
    
    if p[0]== p[0].upper():
        point+=1
    for i in "123456789":
        if i in p:
            point+=1
            break
    if len(p)>=8:
        point+=1
        
    if not any(char.isupper() for char in p):
        point=-2  
        
    if not any(char.islower() for char in p):
        point-=2
        
    if '@' in p or '#' in p or '&' in p:
        point+=2
    return point
    
    
        
```

```python
#tells password good or exellent or time to change the password 
def strength(p):
    if p>=4:
        return "Exellent"
    elif p>=3:
        return "good"
    else:
        return " time Change your password"
```

```python
password = input("enter your pass word ")

point_res = check_points(password)

result = strength(point_res)

print(f"Your password got {point_res} points and its {result}")
```

```python
#Calculate the cost of the cart 

def calculate_total_cost(cart):
    total_cost=0
    
    for item in cart:
        total_cost+=item["price"]*item["quantity"]
    return total_cost
        
    
    


cart=[
    
    {"name":"apple", "price":0.5,"quantity":12},
    {"name":"Banana", "price":0.2,"quantity":6},
    {"name":"Orange", "price":0.4,"quantity":9}
    
]


result = calculate_total_cost(cart)

print(f"You have to pay the bill of {result} $")
```

```python
#check a string is palindrome 

def palindrome(string):
    
    string = string.lower().replace(" ","")
    
    count =0
    i=0
    
    while i<len(string)//2:
        if string[i] != string[(len(string)-1)-i]:
            #print("your string is not palindrome ")
            break
        else:
            count+=1
           
        i+=1
    print(count)
    if count == len(string)//2:
        return "this is a palindrome string"
    else:
        return "oops i think it not a palindrome string"
    
    
            
            
```

```python
naam = input("Enter the string ")

result = palindrome(naam)
print(result)
```

```python
# Calculate factorial using recursion

def fact(num):
    if num==0:
        return 1
    else:
        return num*fact(num-1)
  
    

result = fact(5)
print(result)
```

```python
# afunction to read a file and count the frequency of each word

def count_word(file_path):
    word_count={}
    
    with open(file_path,'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip('.,!?;:"\'')
                word_count[word]=word_count.get(word,0)+1
    return word_count

file_path ='hello.txt'
result = count_word(file_path)
print(result)
            
```
