#file operation -Read and write files 
Python provides built -in functions and methods to read and write to files .

```python
##whole file ->with open read only 
with open("new.txt",mode="r") as file:
    for line in  file:
        words=line.split(" ")
        for word in words:
            print(word.strip(),end=" ")#here strip will delete the new line character
        print()
```

```python
#method 2 for doing the same thing 
with open("new.txt",mode="r") as file:
    content = file.read()
    print(content)
```

```python
#writing a file(overwritting )

with open("new.txt",mode="w") as file:
    file.write('Hello world \n')
    file.write("This is a new line ....")

#only this hello world and this is a new line
#will remain everything else will be 
#overwritten
```

```python
#write a file (without overwritten) append mode

with open("new.txt",'a') as file:
    file.write("\nAppend operation take place")
    file.write("This is new line ")
```

```python
##writing a list of lines to the files 

lst = ["first line\n","Second Line\n","third line\n"]

with open("new.txt", "a") as file:
    file.writelines(lst)

```

```python
#Binary files -> writing
data = b'\x00\x01\x03\x04'

with open("file.bin",'wb') as file:
    file.write(data)
    



```

```python
#reading a binary file
with open("file.bin",'rb') as file:
    content = file.read()
    print(content)
    
```

```python
#creating a file with destin.txt and 
#write the content of new.txt to it

with open("new.txt","r") as file:
    content=file.read()
    
with open("destin.txt","w") as file:
    file.write(content)
```

```python
#read a file and count word ,lines , character

wordy={}
char={}
lines=0

with open("new.txt","r") as file:
    for line in file:
        lines+=1
        words = line.split(" ")
        for word in words:
            wordy[word]=wordy.get(word,0)+1
            ch = [x for x in word]
            for c in ch:
                char[c] = char.get(c,0)+1
                
print(wordy)
print(lines)
print(char)
    
```

```python
##Writing and then reading a file 

with open("new.txt","w+") as file:
    file.write("hello world \n")
    file.write("This is a new line \n")
    #Move the file cursor to the beginning 
    file.seek(0)#seek() will help in this 
    content = file.read()
    print(content)
    
```
