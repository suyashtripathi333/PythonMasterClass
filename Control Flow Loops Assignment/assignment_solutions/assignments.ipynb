```python
#Write a program that calculates the factorial of a number input by the user using a while loop.

num = int(input("enter the  number "))

prod = 1

while num >=1:
    prod = prod * num
    num-=1
print("The factorial of the number is : ", prod)
```

```python
#Write a program that calculates the sum of the digits of a number input by the user using a while loop.

num = int(input("enter the number "))
sum =0

while num > 0:
    rem = num%10
    sum = sum + rem
    num = num//10 # using floor division assuming that both number are integer

print("The sum of the number is : ", sum)    
```

```python
#Write a program that checks if a number input by the user is a prime number using a for loop.
num = int(input("enter the number"))

if num > 1:
    for i in range(2,num):
        if num%i==0:
            print("Number is not prime")
            break
    else:
        print("The number is a prime number")
```

```python
#reverse a number 12345 or 54321

num = int(input("Enter the number to reverse"))

rev =0

while num > 0:
    rem = num % 10
    rev = rev * 10 + rem
    num = num //10
print(f"The reverse of the {num} is {rev}")
```

```python
# To check Palindrome
num = int(input("Enter the number"))

new_num = num
rev =0

while num > 0:
    rem = num % 10
    rev = rev * 10 + rem
    num = num //10
    
print(rev)
if new_num == rev:
    print("The given number is palindrome")
else:
    print("Nope , that does not look like palindrome")




```

```python
# Armstrong number
num = int(input("Enter the  number here"))
n= num
m = num
count =0
sum =0

#first while to count length

while n>0:
    n = n //10
    count+=1

# second part to get the sum of the digit^count

while m>0:
    rem = m%10
    sum = sum +pow(rem,count)
    m = m//10

#Now check for armstrong number   
if num == sum:
    print("this input is armstrong number")
else:
    print("this is not an armstrong number")    
```

```python
#Factorial function
def fact(a):
    prod =1
    while a>1:
        prod = prod * a
        a-=1
    return prod
```

```python
#Strong number
num = int(input("Enter the number "))
#n = num
m = num
sum=0
while m>0:
    rem = m % 10
    sum = sum + fact(rem)
    m = m//10

print(sum)
if num == sum:
    print("The input is Strong number")
else:
    print("The number is not A strong number")
    

```
