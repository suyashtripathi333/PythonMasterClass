```python
#claculator 

try: 
    a=int(input("Enter a number here "))
    b=int(input("Enter the second number here "))
    
    sign = input("enter the sign here ")
    
    if sign == '+':
        c=a+b
    elif sign == '/':
        c= a//b
    elif sign == '*':
        c = a*b
    print(c)
except ZeroDivisionError as ex :
    print(ex)
    print("division by zero is invalid use bigger number ")
except Exception as e:
    print(e)
    print("Learn from the mistakes ")
    
```

```python
#try ,except,else block 
try :
    num = int(input("enter the  number here "))
    result = 10//num
except ZeroDivisionError as ex :
    print(ex)
    print("division by zero is invalid use bigger number ")
except ValueError:
    print("The value give is not proper ")
except Exception as e:
    print(e)
    print("Learn from the mistakes ")
else:
    print(result)
    print("Program executed successfully ")
```

```python
#try ,except,else and finally 

try :
    num = int(input("enter the  number here "))
    result = 10//num
except ZeroDivisionError as ex :
    print(ex)
    print("division by zero is invalid use bigger number ")
except ValueError:
    print("The value give is not proper ")
except Exception as e:
    print(e)
    print("Learn from the mistakes ")
else:
    print(f"The program executed the result is {result}")
finally:
    print("Execution complete. ")
    
```

```python
##file handling and Exception handling 

try:
    file = open("new.txt","r") 
    content = file.read()
except FileNotFoundError as ex:
    print(ex)
else:
    print(content)
finally:
    if 'file' in locals() and not file.closed:
        file.close()
        print("file is closed ")
        
    
```
